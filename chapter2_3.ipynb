```python
# WAP to print all even & odd numbers separately from 1 to 50 using a loop and conditional statements.

even_number=[]
odd_number=[]

for i in range(1,51):
    if i%2==0:
        even_number.append(i)
    else:
        odd_number.append(i)

print("Even numbers from 1 to 50 :",even_number)
print("Odd numbers from 1 to 50 :",odd_number)
```

```python
# WAP to find the sum of all numbers between 1 and 100 that are divisible by both 3 and 5.

sum=0
for i in range(1,101):
    if i%3==0 and i%5==0:
        sum=sum+i
print(sum)
```

```python
# WAP to accept 10 numbers from the user and count how many are positive, negative, and zero using a loop and conditional statements.

num=[]
for i in range(1,11):
    a=int(input("Enter a number :"))
    num.append(a)

for i in num:
    if i>0:
        print("Positive")
    elif i<0:
        print("Negative")
    else:
        print("Zero")
```

```python
# WAP to check whether a given number is an Armstrong number or not using a loop and conditional statements.

num=int(input("Enter a number :"))
a=num
b=len(str(a))
arm=0

while num > 0:
    digit=num%10 # last digit
    arm=arm+digit**b
    num=num//10

if a==arm:
    print("Armstrong number")
else:
    print("Not Armstrong number")
    
```

```python
# WAP using nested loops to print all pairs (i, j) where i and j range from 1 to 5, but display only those pairs whose sum is even.

for i in range(1,5):
    for j in range(1,5):
        if (i+j)%2==0:
            print((i,j))
```

```python
# WAP using nested loop to print the multiplication tables from 2 to 5, with each table containing multiples from 1 to 10. Use a conditional statement to display only the multiples that are divisible by 3.

for num in range(2,6):
    for i in range(1,11):
        if num*i<=10 and num*i%3==0:
            print(num ,"x" ,i ,"=" ,num*i)
```

```python
# WAP to print numbers from 1 to 20, but use continue to skip all numbers that are divisible by 3.

for i in range(1,21):
    if i%3==0:
        continue
    print(i)
    
```

```python
# WAP that repeatedly accepts numbers from the user and calculates their sum. Terminate the loop using break when the user enters 0.

num=1
while num != 0:
    a=int(input("Enter a number :"))
    if a==0:
        break
```
