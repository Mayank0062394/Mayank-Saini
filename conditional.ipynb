```python
# Enter student details:
#     * If marks ≥ 85 → Academic scholarship
#     * If sports = yes → Sports scholarship
#     * If income < ₹2,00,000 → Financial aid
#     (Student may get multiple benefits.)

marks=float(input("Enter student's marks"))
sport=input("Sports scholarship wanted or not(yes/no)")
income=float(input("Enter student's income"))

if marks>=85:
    print("Academic scholarship")
if sport=="yes" sport=="Yes":
    print("Sports scholarship")
if income<200000:
    print("Financial aid")
else:
    print("Error")
```
